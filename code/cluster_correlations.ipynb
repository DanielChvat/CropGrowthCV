```python
import pandas as pd
import numpy as np
```

```python
drop_cols = ['OID_', 'SOURCE_ID', 'IS_SEED']
all_june_20 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_june_20_3c.csv").drop(columns=drop_cols)
all_july_20 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_july_20_3c.csv").drop(columns=drop_cols)

all_june_21 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_june_21_3c.csv").drop(columns=drop_cols)
all_july_21 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_july_21_3c.csv").drop(columns=drop_cols)

all_june_22 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_june_22_3c.csv").drop(columns=drop_cols)
all_july_22 = pd.read_csv("../Datasets/Clusters/3_cluster_spatial/all_july_22_3c.csv").drop(columns=drop_cols)
```

```python
def ClusterCorrelation(df1, df2):
    correlations = []
    n_clusters = len(np.unique(df1['CLUSTER_ID']))
    features = [col for col in df1.columns if col != 'CLUSTER_ID']
    for i in range(1, n_clusters + 1):
        corr_df1 = df1[df1['CLUSTER_ID'] == i].reset_index(drop='True')
        for j in range(1, n_clusters + 1):
            corr_df2 = df2[df2['CLUSTER_ID'] == j].reset_index(drop='True')
            correlations.append(np.average(corr_df1[features].corrwith(corr_df2[features]).values))

    return correlations
```

### Correlation between June 2020, June 2021, and June 2022

```python
cols = ['2021 Cluster ' + str(i) for i in range(1,4)] + ['2022 Cluster ' + str(i) for i in range(1,4)]
df = pd.DataFrame(columns=['2020 Cluster'] + cols)
corr_2020_2021 = ClusterCorrelation(all_june_20, all_june_21)
corr_2020_2022 = ClusterCorrelation(all_june_20, all_june_22)
df.loc[len(df)] = ['1'] + corr_2020_2021[:3] + corr_2020_2022[:3]
df.loc[len(df)] = ['2'] + corr_2020_2021[3:6] + corr_2020_2022[3:6]
df.loc[len(df)] = ['3'] + corr_2020_2021[6:9] + corr_2020_2022[6:9]
df.set_index("2020 Cluster", inplace=True)
df
```

### Correlation between June 2021 and June 2022

```python
cols = ['2022 Cluster ' + str(i) for i in range(1,4)]
df = pd.DataFrame(columns=['2021 Cluster'] + cols)
corr_2021_2022 = ClusterCorrelation(all_june_21, all_june_22)
df.loc[len(df)] = ['1'] + corr_2021_2022[:3]
df.loc[len(df)] = ['2'] + corr_2021_2022[3:6]
df.loc[len(df)] = ['3'] + corr_2021_2022[6:9]
df.set_index("2021 Cluster", inplace=True)
df
```
