```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neural_network import MLPRegressor
from xgboost import XGBRegressor
from IPython.display import display

from Preprocess import Preprocessor
from GridCV import GridCV
```

#### Load Field Indices, Set Split Method, and Define Models to be Tested Through Cross Validation

```python
indices_6302020 = pd.read_csv("../Datasets/Indices_Combined/2020/June_30_2020_New.csv")
indices_6302021 = pd.read_csv("../Datasets/Indices_Combined/2021/June_30_2021.csv")
indices_6282022 = pd.read_csv("../Datasets/Indices_Combined/2022/June_28_2022.csv")

indices_7062020 = pd.read_csv("../Datasets/Indices_Combined/2020/July_6_2020_New.csv")
indices_7122021 = pd.read_csv("../Datasets/Indices_Combined/2021/July_12_2021.csv")
indices_7132022 = pd.read_csv("../Datasets/Indices_Combined/2022/July_13_2022.csv")

split_method = 'representative'
cv_models = {'LR': LinearRegression(n_jobs=-1), 'RF': RandomForestRegressor(n_jobs=-1, random_state=41, max_features='sqrt', max_depth=6, max_samples=50), 'XGB': XGBRegressor(eta=0.02, max_depth=4, subsample=0.5), 'FFN': MLPRegressor(random_state=41, batch_size=100, early_stopping=True, learning_rate='adaptive')}
```

### June 2020, 2021, 2022

```python
p = Preprocessor()

X_train_validate, Y_train_validate = p.transform([indices_6302020, indices_6302021, indices_6282022])

folds = GridCV.split([indices_6302020, indices_6302021, indices_6282022], n_splits=12, method=split_method)

grid_validator = GridCV(models=cv_models)

grid_results = grid_validator.results(X_train_validate, Y_train_validate, folds)

display(grid_results)
```
